### 1. Load 3 Datasets (MNIST, Fashion-MNIST, UCI Digits)

```
datasets = load_all_three_datasets('../data', batch_size=64)
print('Dataset 1 (MNIST): 60,000 train, 10,000 test (28x28x1)')
print('Dataset 2 (Fashion-MNIST): 60,000 train, 10,000 test (28x28x1)')
print('Dataset 3 (UCI Digits): 1,797 samples (64 features)')
```

### 2. Discover CNN: Shape Inspection & Parameter Counting

```
pt_model = PyTorchCNN(in_channels=1, num_classes=10, spatial_size=28)
dummy_x = torch.randn(2, 1, 28, 28)
shapes = pt_model.inspect_shapes(dummy_x)
for name, s in shapes:
    print(f'{name}: {s}')
_, total_params = pt_model.count_parameters_detailed()
print(f'Total trainable parameters: {total_params:,}')
```

### 3. CNN from Scratch (Pure NumPy)

```
raw_m_train, raw_m_test = datasets['raw_mnist']
X_tr = (raw_m_train.data.numpy()[:2500, None, :, :].astype(np.float32)/255.0 - 0.1307) / 0.3081
y_tr = raw_m_train.targets.numpy()[:2500]
X_va = (raw_m_train.data.numpy()[2500:3000, None, :, :].astype(np.float32)/255.0 - 0.1307) / 0.3081
y_va = raw_m_train.targets.numpy()[2500:3000]
X_te = (raw_m_test.data.numpy()[:1000, None, :, :].astype(np.float32)/255.0 - 0.1307) / 0.3081
y_te = raw_m_test.targets.numpy()[:1000]

scratch_net = CNN_Scratch(in_channels=1, num_classes=10)
hist_np, t_np = scratch_net.fit(X_tr, y_tr, X_va, y_va, epochs=5, batch_size=64, lr=2e-3)
preds_np = scratch_net.predict(X_te)
print('NumPy Metrics:', evaluate_metrics(y_te, preds_np))
```
